```python
import json

# Đọc file JSON gốc
with open(r"D:\labeled_queries1.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# Xử lý từng query
for item in data:
    seen = set()
    unique_relevant = []
    for rel in item["relevant"]:
        lawid = rel["lawid"]
        if lawid not in seen:
            # Tạo bản sao mới nhưng bỏ trường similarity_score nếu có
            new_rel = {k: v for k, v in rel.items() if k != "similarity_score"}
            unique_relevant.append(new_rel)
            seen.add(lawid)
    item["relevant"] = unique_relevant

# Ghi lại file JSON mới
with open("data_query_fix.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print("✅ Đã xóa các lawid trùng và bỏ cột similarity_score thành công!")
```
